```python
# Enumerate

import numpy as np
arr = np.array([[12, 20, 32, 15], [16, 52, 12, 23], [16, 50, 45, 78]])

for idx, x in np.ndenumerate (arr):
  print ((idx, x))

# numpy.ndenumerate is a multidimensional index iterator that returns an iterator yielding pairs of array coordinates (as a tuple) and the corresponding element value
# numpy.ndenumerate() lets you go through an array and get both the position of each element and its value.

arr_ = np.array ([12, 24, 36, 48, 56, 84])

for idx, x in np.ndenumerate (arr_):
  print (idx, x)

arr__ = np.array ([1, 2, 2, 4, 3, 6, 4, 8, 5, 6, 8, 4])

for idx, x in np.ndenumerate (arr__):
  print (idx, x)

# Non integers can by no means be Concatenated
#arr_3 = np.concatenate (arr_, arr__)
#print (arr_3)
```

```python
import numpy as np

arr1 = np.array([[1, 2], [3, 4], [23, 16]])

arr2 = np.array([[5, 6], [7, 8]])

arr = np.concatenate((arr1, arr2), axis= None)
# None here is used to define that we are merging an array with different number of Dimensions

print(arr)

# Or Using stack functions
# NB: Using np.stack we cannot join arrays of different dimensions.

arr_1 = np.array([[1, 2], [3, 4], [23, 16]])

arr_2 = np.array([[5, 6], [7, 8], [19, 17]])

arr_3 = np.stack ((arr_1, arr_2), axis= 1)
print (arr_3)

# Stacking along rows
arr_3 = np.hstack ((arr_1, arr_2))
print (arr_3)

# Stacking along columns
arr_3 = np.vstack ((arr_1, arr_2))
print (arr_3)

# Stacking Along height (depth)
arr_3 = np.dstack ((arr_1, arr_2))
print (arr_3)
```
