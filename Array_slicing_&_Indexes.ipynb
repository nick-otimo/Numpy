```python
import numpy as np
from numpy import random
import matplotlib.pylab as plt


list_1 = np.random.randint (5, 47, 6)
print (f"List one: {list_1}")

# Change the values at index
list_1 [-1] = 23
list_1 [0] = 10
print (f"Modified List one: {list_1}") # To index the last object in a list use -1

# 2D array
list_2 = np.random.randint (5, 17, size = (4, 5))
print ("List two:")
print()
print (list_2)
list_2 [0, 1] = 15
print()
print ("Modified List two:")
print (list_2)

list_2_avg = np.average (list_2)
print (f"Average of List two: {list_2_avg}")
print (f"The shape of List two: {list_2.shape}")
print (f"The size of List two: {list_2.size}")
# Shape - Rows, columns
# Size - Rows * columns

# Get the value by the index
print (f"The value is: {list_2[1, 2]}")
print (f"The item value is: {list_2.item (1, 2)}")

# Get the specific indices

list_3 = np.take (list_2, [0, 2, 10, 16])
print (f"The specific indices are: {list_3}")
```

```python
import numpy as np
from numpy import random
import matplotlib.pylab as plt
list_1 = np.random.randint (1, 15, size = (4, 5))

print ("Below is our List 1")
print ("==" * 9)
print (list_1)
print ("==" * 9)

# Flip array
# fliped_array = np.flip (list_1)
# print (fliped_array)
print ("The flipped array is:")
print()
print ( list_1[::-1])

# Get the values which are greater than a particular number
print ()
print (f"These values of list one are greater than 7: {list_1 [list_1 > 7]}")
print (f"These values of list one are less than 3: {list_1 [list_1 < 4]}")
print (f"These values of list one are Equal to 5: {list_1 [list_1 <= 9]}")

# Get the numbers between 5 and 12
btn_5_12 = list_1 [(list_1 > 5) & (list_1 < 12)]
print (f"The numbers between 5 and 12: {btn_5_12}")
avg_btn_5_12 = np.average (btn_5_12)
print (f"The average of these numbers: {avg_btn_5_12}")


# Get the unique numbers

list_1_unq = np.unique (list_1)
print (f"The unique numbers are: {list_1_unq}")

```
