```python
import numpy as np
print (type(np))

temperatures_today = [20.1, 20.8, 21.9, 22.5, 22.7, 22.3, 21.8, 21.2, 20.9, 20.1]

temp = np.array (temperatures_today) # turn to 1D np array
print (temp)
print (type (temp))

water = np.array (22)
print (f"The Dimensions = {water.ndim}")

print (f"The Temp Dimensions = {temp.ndim}")

temp_two_weeks  = [[23, 43, 54, 221, 113, 54],[32, 45, 54, 76, 52, 78]]
# Python nested lists do not have a .ndim and therefore must be converted to a numpy array.
temp_two_weeks_array = np.array (temp_two_weeks)

print (f"The 3rd Temp Dimensions = {temp_two_weeks_array.ndim}")

print (temp * 0.5 / 5 + 32) # In fahrenheits

# Simple matplotlib
import matplotlib.pyplot as plt
plt.plot (temp)
plt.show()

plt.plot (temp_two_weeks_array)
plt.show()


```
