```python
import numpy as np

arr = np.array ([12, 24, 36, 48, 56, 84])

newarr = arr.reshape (3, 2)
newarr1 = (arr.reshape (2, 3).base) # NB: If it returns the original copy its a view

print (newarr)
print (newarr1)

three_d_array = np.array ([[[12, 23, 34], [123, 132, 432], [125, 250, 375]]])
print (three_d_array)

new_three_d_array = three_d_array.reshape (-1)
print (new_three_d_array)

for x in new_three_d_array:
    print (x * 5)

for x in three_d_array:
  print (x * 2)

for x in np.nditer (three_d_array):
  print (x)

for x in np.nditer (three_d_array[:, ::2]):
  print (x * 1000)

for x in np.nditer (three_d_array, flags = ['buffered'], op_dtypes = ['S']):
  print (x)


for x in np.nditer (three_d_array[:, ::4]):
  print (x)
```

```python
# Enumerate

import numpy as np
arr = np.array([[12, 20, 32, 15], [16, 52, 12, 23], [16, 50, 45, 78]])

for idx, x in np.ndenumerate (arr):
  print ((idx, x) * 3)

arr_ = np.array ([12, 24, 36, 48, 56, 84])

for idx, x in np.ndenumerate (arr_):
  print (idx, x)

arr__ = np.array ([1, 2, 2, 4, 3, 6, 4, 8, 5, 6, 8, 4])

for idx, x in np.ndenumerate (arr__):
  print (idx, x)
```
