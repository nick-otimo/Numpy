```python
import numpy as np
arr = np.array (["Maggy", "Paul", "Nick", "ELvis", "Velma", "Blessing"])

print (arr.dtype)

arr_two = np.array ([12, 23, 43, 54, 65])
print (arr_two.dtype)

arr_three = np.array (["1.0", "2.2", "5.4", "4.1", "1.5"])
print (arr_three.dtype)

# Create an array with the data type string [S]

arr_four = np.array ([9, 7, 5, 3, 5, 2, 1], dtype= 'S')

print (arr_four)
print (arr_four.dtype)
```
