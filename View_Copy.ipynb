```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.pylab as plt
from numpy import random

list_1 = [12, 1, 3, 6, 7, 9]
array_list_1 = np.array (list_1)
print (f"My array is: {array_list_1}")

# Multidimensional array
list_2 = [[[1, 2, 3], [4, 5, 6], [7, 8, 9]]]
array_list_2 = np.array (list_2)
print (f"My another multidimensional array is: {array_list_2.ndim}")

numbers = np.arange (11, 20, 2)
print (f"My numbers are: {numbers}")

numbers2 = numbers.copy()
print (f"Numbers 2: {numbers2}")

numbers2 [3] = 40
print (numbers2)

print (max(numbers2))
print (min(numbers2))
total = min (numbers2) * max (numbers2)
print (total * 415)

number = range (4, 12)
print (f"My other numbers are: {list(number)}")

number3 = np.copy(number)
print (f"My Number 3 is: {number3}")
number3 [3] = 14 * 7
print (f"My Number 3 is: {number3}")
average = np.average (number3)
print (f"My average is: {average}")




```

```python
import numpy as np
# generates an array of (x) evenly spaced floating-point numbers starting at 0.0 and ending at 5.0 (inclusive)
# x = 10 here.
arr = np.linspace(0, 5, 10)
print (arr)

# Create a 3 item arrays of Zeros
arr_2 = np.zeros (7)
print (arr_2)

# you can create multidimensional arrays of zeros by passing
# a tuple with th 1st value being rows and the 2nd columns
arr_3 = np.zeros((4, 3))
print (f"Array three: {arr_3}")

# Create array of 0nes
arr_4 = np.ones (5)
print (f"My 1D array of ones: {arr_4}")

arr_5 = np.ones ((5, 3))
print (f"My 2D array of ones: {arr_5}")
```
