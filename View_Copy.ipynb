```python
import numpy as np
numbers = np.arange (11, 20, 2)
print (f"My numbers are: {numbers}")

numbers2 = numbers.copy()
print (f"Numbers 2: {numbers2}")

numbers2 [3] = 40
print (numbers2)

print (max(numbers2))
print (min(numbers2))
total = min (numbers2) + max (numbers2)
print (total * 415)

number = range (4, 12)
print (f"My other numbers are: {list(number)}")

number2 = np.copy(number)
print (f"My Number 2 is: {number2}")
number2 [3] = 14 * 7
print (f"My Number 2 is: {number2}")
average = np.average (number2)
print (f"My average is: {average}")




```
